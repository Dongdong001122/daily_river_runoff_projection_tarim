```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

station_names = ['Huangshuigou', 'Dashankou', 'Xiehela', 'Shaliguilanke', 'Kaqun', 'Wuluwati', 'Tongguziluoke']
stations = ['hsg', 'dsk', 'xhl', 'slglk', 'kq', 'wlwt', 'tgzlk']
ssps = ['ssp126', 'ssp245', 'ssp370', 'ssp585']

fig, axes = plt.subplots(7, 1, figsize=(8.27, 11.69), sharey=True)  # Single subplot per station

bar_width = 2.5  # Adjust bar width for clarity
colors = ['b', 'g', 'r', 'y']  # Colors for different SSPs

for i, station in enumerate(stations):
    ax = axes[i]  # Single axis for each station

    for j, ssp in enumerate(ssps):
        file_path = f"F:/geodata/river_runoff_obs/output/{stations.index(station)+1}_imputMF_{ssp}_daily_30_fixed_mean.csv"
        df = pd.read_csv(file_path, index_col=0)

        df['time_daily'] = pd.to_datetime(df.index)
        quantile_95 = df['runoff_fixed'].quantile(0.95)
        high_runoff_days = df[df['runoff_fixed'] > quantile_95]
        high_runoff_days['year'] = high_runoff_days['time_daily'].dt.year
        days_above_95_quantile = high_runoff_days.groupby('year').size()

        days_above_95_quantile.index = pd.to_datetime(days_above_95_quantile.index, format='%Y')
        decadal_data = days_above_95_quantile.resample('10Y').sum()

        x_positions = decadal_data.index.year + (j - 1.5) * bar_width  # Offset bars for each SSP
        ax.bar(x_positions, decadal_data.values, width=bar_width, label=ssp.upper(), color=colors[j])

    ax.set_title(f'{station_names[i]}', fontsize=10)
    ax.set_xlabel('Year', fontsize=10)
    ax.set_ylabel('Number of High Runoff Days', fontsize=10)
    ax.set_xlim(2015, 2100)
    ax.legend()  # Add legend to differentiate SSP scenarios

plt.tight_layout()
plt.savefig(r'F:\geodata\river_runoff_obs\high_flow_frequency.svg')
plt.show()
```
