```python
arcpy.env.mask ="xinjiang_61_"
```

```python
# Define your input raster and target area
input_raster = f'F:\geodata\Desert\desertProject\DDI2015.tif'
target_area = 31877.0  # Target area in square units
```

```python
def calculate_raster_sum(raster_path):
    """
    Calculates the sum of all pixel values in a raster.
    
    :param raster_path: The path to the input raster file.
    :return: The sum of all pixel values as a float.
    """
    try:
        # Load the raster using the Raster object
        raster = arcpy.Raster(raster_path)
        
        # Convert the raster to a NumPy array
        raster_array = arcpy.RasterToNumPyArray(raster)
        
        # Calculate the sum of all pixel values
        raster_sum = raster_array.sum()
        
        # Convert the sum to float
        return float(raster_sum)
    
    except Exception as e:
        print(f"Error calculating raster sum: {e}")
        return None
```

```python
DDI_raster = Raster(input_raster)
DDI_min = DDI_raster.minimum
```

```python
DDI_low_area = 0 # This is the area of DDI lower than the interval
area_interval = DDI_min # This is the interval for reclassify the DDI raster
while -DDI_low_area+target_area > 100: # 100 is the tolerance
    DDI_low = DDI_raster<=area_interval

    DDI_low_raster_info = DDI_low.computeStatistics()[0]
    DDI_low_area=(0.03*0.03)*DDI_low_raster_info['mean'] * DDI_low_raster_info['count']
    
    if target_area - DDI_low_area >2000:
        area_interval+=0.2
    else:
        area_interval += 0.05
    print(area_interval,DDI_low_area)
    
DDI_low.save("DDI_low_2015")    
```
