```python
import pandas as pd
```

```python
name = '3_xhl_imputMF_future'
df = pd.read_csv(f"F:\\geodata\\river_runoff_obs\\{name}.csv")
```

```python
df.tail()
```

```python
df.columns
```

```python
# Set 'time' column as index if it isn't already
df['time'] = pd.to_datetime(df['time'])  # Ensure time is datetime
df.set_index('time', inplace=True)       # Set it as the index

# Resample from daily to monthly
resampled_df = df.resample('M').agg({
    'pre': 'sum',    # Sum for 'pre'
    'dis': 'sum',    # Sum for 'dis'
    'tm': 'mean',    # Mean for 'tm'
    'tmax': 'mean',  # Mean for 'tmax'
    'tmin': 'mean'   # Mean for 'tmin'
})
```

```python
resampled_df.tail()
```

```python
resampled_df.to_csv(f"F:\\geodata\\river_runoff_obs\\{name}_month.csv")
```

```python
# Resample from daily to monthly
resampled_10d_df = df.resample('10D').agg({
    'pre': 'sum',    # Sum for 'pre'
    'dis': 'sum',    # Sum for 'dis'
    'tm': 'mean',    # Mean for 'tm'
    'tmax': 'mean',  # Mean for 'tmax'
    'tmin': 'mean'   # Mean for 'tmin'
})
resampled_10d_df
```
