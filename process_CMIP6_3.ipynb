# This code script is for multimodel ensembles

```python
import os
import pandas as pd
import glob
os.chdir('F:\\geodata\\river_runoff_obs')
```

```python
def extract_metadata(file_name):
    parts = file_name.split('_')
    station = parts[1]
    gcm = parts[2]
    scenario = parts[3]
    return station, gcm, scenario
```

```python
file_name_list = ['1_hsg_imputMF','2_dsk_imputMF','3_xhl_imputMF','4_slglk_imputMF','5_kq_imputMF','6_wlwt_imputMF','7_tgzlk_imputMF']
```

```python
for file_name in file_name_list:
    abbre = file_name.split('_')[1]
    for scenario in ['ssp126','ssp245','ssp370','ssp585']:
        print(abbre,scenario)
        csv_daily_files = glob.glob(f'{file_name}*{scenario}*daily.csv')
        csv_daily_files = [item for item in csv_daily_files if 'multi-model' not in item]
        daily_df_list = [pd.read_csv(file,index_col=0) for file in csv_daily_files]
        daily_df_mean = sum(daily_df_list)/len(daily_df_list)
        daily_df_mean['dis'] = daily_df_list[0]['dis']
        daily_df_mean.to_csv(f'{file_name}_multi-model_{scenario}_r1i1p1f1_daily.csv')
        
        csv_mon_files = glob.glob(f'{file_name}*{scenario}*mon.csv')
        csv_mon_files = [item for item in csv_mon_files if 'multi-model' not in item]
        mon_df_list = [pd.read_csv(file,index_col=0) for file in csv_mon_files]
        mon_df_mean = sum(mon_df_list)/len(mon_df_list)
        mon_df_mean['dis'] = mon_df_list[0]['dis']
        mon_df_mean['gr'] = mon_df_list[0]['gr']
        
        daily_df_mean.to_csv(f'{file_name}_multi-model_{scenario}_r1i1p1f1_daily.csv')
        mon_df_mean.to_csv(f'{file_name}_multi-model_{scenario}_r1i1p1f1_mon.csv')
        
        print(mon_df_mean)
```
