## This script is for read the RGI dbf file recoding the basins need to be validated 

```python
import pandas as pd
from dbfread import DBF
from simpledbf import Dbf5
```

```python
# Path to the .dbf file
dbf_path = r"C:\Users\HP\Documents\ArcGIS\Projects\LSTM_pre0\c13_rgi60_SpatialJoin.dbf"
dbf = Dbf5(dbf_path)

# Convert the .dbf file to a pandas DataFrame
RGI6_df  = dbf.to_dataframe()
RGI6_df.set_index('RGIId', inplace=True)
```

```python
RGI6_df
```

```python
for scenario in ["ssp" + str(x) for x in [ 126,245, 370, 585]]:
    print(scenario)
    peak_water_df = pd.read_csv(f"I:\\GlacierData\\peakwater\\R131415_glac_runoff_fixed_monthly_1set_2000_2100-{scenario}-Batch.csv", index_col=0)
    
    peak_water_df=peak_water_df[['BCC-CSM2-MR', 'CESM2', 'CESM2-WACCM', 'EC-Earth3', 'EC-Earth3-Veg', 'FGOALS-f3-L', 'GFDL-ESM4', 'INM-CM4-8', 'INM-CM5-0', 'MPI-ESM1-2-HR', 'MRI-ESM2-0', 'NorESM2-MM']].apply(pd.to_numeric, errors='coerce').astype('Int64')
    peak_water_df
    peak_water_df[scenario] = peak_water_df.median( axis=1)
    RGI6_df = RGI6_df.merge(peak_water_df[scenario], how='left', left_index=True, right_index=True)
```
